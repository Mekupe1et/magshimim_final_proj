```python
import pandas as pd
base = pd.read_csv("../base_dbs/twitter.csv")
base.head()
```

```python
temp = base.copy()
```

```python
temp = temp[temp["Definitely English"] == 1]
temp
```

```python
temp = temp[['Tweet', 'Date']]
temp
```

```python
temp["year"] = temp["Date"].str.split("-", expand=True)[0].astype(int)
temp["decade"] = (temp["year"] // 10) * 10
temp.drop(['Date'], axis=1, inplace=True)
temp.rename(columns={"Tweet": "text"}, inplace=True)
temp.head()
```

```python
temp["category"] = "tweet"
temp.head()
```

```python
df = pd.read_parquet("../large_dataset.parquet")
df = pd.concat([df, temp], ignore_index=True)
df.to_parquet("large_dataset.parquet")
df.head()
```
