```python
import pandas as pd
df = pd.read_csv("../base_dbs/data.csv")
```

```python
df = df[["datetime","text"]]
```

```python
df["datetime"] = pd.to_datetime(df["datetime"])
```

```python
df['year'] = df['datetime'].dt.year
```

```python
df.drop("datetime",inplace=True,axis=1)
```

```python
df["year"].value_counts()
```

```python
df_sample = df.sample(n=20000, random_state=42).reset_index(drop=True)
```

```python
df_sample.rename(columns={'year': 'decade'}, inplace=True)
```

```python
df_sample["decade"] = 2020
```

```python
df_sample.to_csv("../base_dbs/data.csv", index=False)
```
