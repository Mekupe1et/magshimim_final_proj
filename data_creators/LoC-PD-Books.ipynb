https://huggingface.co/datasets/storytracer/LoC-PD-Books

```python
import pandas as pd
temp = None
for i in range(1,8):
    df = pd.read_parquet(f"base_dbs/LoC-PD-Books/train_0000{i}.parquet")
    df["category"] = "book"
    temp = pd.concat([temp, df], ignore_index=True)
temp
```

```python
temp = temp[['text', 'year', 'category']]
temp["decade"] = (temp["year"] // 10) * 10
temp
```

```python
df = pd.read_parquet("../large_dataset.parquet")
df = pd.concat([df, temp], ignore_index=True)
df.to_parquet("large_dataset.parquet")
df.head()
```
