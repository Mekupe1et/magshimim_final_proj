```python
import pandas as pd
temp = pd.read_csv("../base_dbs/textage_large.csv")
temp.head()
```

```python
temp["year"] = temp["date"].astype(int)
temp.drop(['title', 'text_length', 'author', 'date'], axis=1, inplace=True)
temp.head()
```

```python
df = pd.read_parquet("../large_dataset.parquet")
df = pd.concat([df, temp], ignore_index=True)
df.to_parquet("large_dataset.parquet")
df.head()
```
