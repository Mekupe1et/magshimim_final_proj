```python
import pandas as pd
base = pd.read_csv("../base_dbs/youTube.csv")
base.head()
```

```python
base = base[base["language"] == "en"]
base["year"] = base["published_at"].str.split("-", expand=True)[0].astype(int)
base["text"] = base["comment_text"]
base = base[['text', 'year']]
base["decade"] = (base["year"] // 10) * 10
base["category"] = "youtube comment"
base["language_type"] = "written"
base
```
