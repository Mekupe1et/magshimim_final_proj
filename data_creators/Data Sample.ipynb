```python
import pandas as pd
df = pd.read_parquet("../processed_dbs/combined_dataset.parquet")
```

```python
filtered_df = df[df["category"] != "None"]

new_df = (
    filtered_df.groupby("decade", group_keys=False)
    .apply(
        lambda x: x.sample(n=min(20000, len(x)), random_state=42),
        include_groups=False,
    )
    .reset_index(drop=True)
)
```

```python
df = df[df["decade"] > 1800]
```

```python
new_df = df[df["decade"] == 1]
new_df
```

```python
for i in df["decade"].unique():
    temp = df[(df["decade"] == i) & (df["category"] != "None")].sample(n=20000, random_state=42).reset_index(drop=True)
    new_df = pd.concat([new_df, temp], ignore_index=True)
```

```python
new_df["decade"].value_counts()
```

```python
Sample_Data = new_df
```

```python
Sample_Data.to_parquet("../processed_dbs/Sampled_data.parquet")
```
